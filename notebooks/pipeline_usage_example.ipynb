## 5. Programmatic Configuration

```python
from dualdocking.pipeline.config import ScorerConfig

# Create custom configuration
custom_config = PipelineConfig(
    project_root=Path.cwd().parent,
    pairs=runner.config.pairs,  # Use pairs from file
    modes=["single_a", "single_b", "double"],
    seeds=[42],
    scorer=ScorerConfig(
        engine="boltz2",
        backend="cli",  # Use CLI instead of NIM
    ),
    parallel=1,
    resume=True,
)

custom_runner = PipelineRunner(custom_config)
print(f"Custom config created with {len(custom_config.seeds)} seed(s)")
```

## 6. Generate Task List

```python
# Generate all tasks for a specific pair
tasks = runner.config.generate_tasks(
    pair_filter="CDK2_ERK2",
    seed_filter=[42],
)

print(f"Generated {len(tasks)} tasks:")
for task in tasks:
    print(f"  {task}")
    print(f"    Output dir: {task.run_dir(runner.config.project_root)}")
```
